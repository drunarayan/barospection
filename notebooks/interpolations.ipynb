## For Interpolations

```python
    import numpy as np

    x_values = np.array([0, 1, 2, 3, 4])
    y_values = np.array([10, 12, 15, 13, 18])

    x_interp = 2.5

    y_interp = np.interp(x_interp, x_values, y_values)
    print(y_interp)  # Output: 14.0
```

```python
    from scipy.interpolate import interp1d
    import numpy as np

    x_values = np.array([0, 1, 2, 3, 4])
    y_values = np.array([10, 12, 15, 13, 18])

    x_interp = 2.5

    f = interp1d(x_values, y_values, kind='cubic')
    y_interp = f(x_interp)
    print(y_interp) # Output: 14.21875
```
