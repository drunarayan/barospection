This provides the star celestial coordinates in sexagesimal format to place the star in the center of the image square. However, this is not what we want.  We want to move the star to the left of the image frame so that there will be room for the spectrum to its right. To do this, we are going to add 3.5 arc-min in RA and subtract 3.5 arc-min in Dec. 

To do this properly, we need to convert both the RA and Dec values to a “decimal degrees” format.  

You should map the 11h  of RA hr to 11\*360/24 = 165 deg PLUS 49m  of RA to (49/60)\*(360/24) PLUS 3.57834s of of RA to (3.5784/(60\*60)\*(360/24) giving a value of 177.2625 in RA decimal degrees. This is because 24 hours maps to 360 deg.

You should map the +14° of Dec to 14 deg PLUS 34′ of Dec to 34/60 PLUS 19.4090″ of Dec to 19.4090/3600 giving a value of +14.5720 in Dec decimal degrees.

So an RA of 11h 49m 03.57834s is 177.2625 decimal deg
A Dec of +14° 34′ 19.4090″ is +14.5720 decimal deg

The last step in this process if to add 3.5 arc-min to RA and subtract 3.5 arc-min from Dec giving a final result of adjusted RA & Dec for your star.
