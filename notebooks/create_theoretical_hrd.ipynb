```python
import numpy as np
import matplotlib.pyplot as plt

def create_hrd_plot(star_data):
    """
    Creates and displays an HRD with stellar types labeled.

    Args:
        star_data (dict): A dictionary containing stellar data with keys
                          'temperature' and 'luminosity' or 'absolute_magnitude'.
                          This can be replaced with actual astronomical data.
    """
    # Create the plot
    fig, ax = plt.subplots(figsize=(10, 8))

    # --- Plot the stellar data ---
    # The x-axis is temperature, but it's conventionally reversed for HRDs
    ax.scatter(star_data['temperature'], star_data['luminosity'], s=2, color='black', alpha=0.5, zorder=2)

    # --- Add annotations for different stellar types ---

    # Main Sequence
    ax.text(25000, 1e4, 'O-Type', ha='center', va='bottom', fontsize=10, color='royalblue', fontweight='bold')
    ax.text(10000, 1e2, 'A-Type', ha='center', va='bottom', fontsize=10, color='dodgerblue', fontweight='bold')
    ax.text(7000, 1e1, 'F-Type', ha='center', va='bottom', fontsize=10, color='lightblue', fontweight='bold')
    ax.text(6000, 1, 'G-Type', ha='center', va='bottom', fontsize=10, color='yellow', fontweight='bold')
    ax.text(5000, 1e-1, 'K-Type', ha='center', va='bottom', fontsize=10, color='orange', fontweight='bold')
    ax.text(3500, 1e-2, 'M-Type', ha='center', va='bottom', fontsize=10, color='tomato', fontweight='bold')
    
    ax.text(6000, 1, 'Main Sequence', ha='center', va='bottom', fontsize=14, color='white',
            bbox=dict(facecolor='black', alpha=0.6, boxstyle='round,pad=0.2'), transform=ax.transData)

    # Giants
    ax.text(5000, 1e3, 'Giants', ha='center', va='bottom', fontsize=14, color='white',
            bbox=dict(facecolor='black', alpha=0.6, boxstyle='round,pad=0.2'))
            
    # Supergiants
    ax.text(6000, 1e5, 'Supergiants', ha='center', va='bottom', fontsize=14, color='white',
            bbox=dict(facecolor='black', alpha=0.6, boxstyle='round,pad=0.2'))

    # White Dwarfs
    ax.text(15000, 1e-3, 'White Dwarfs', ha='center', va='bottom', fontsize=14, color='white',
            bbox=dict(facecolor='black', alpha=0.6, boxstyle='round,pad=0.2'))

    # --- Plot stylistic enhancements ---
    ax.set_yscale('log')  # Luminosity axis is logarithmic
    ax.invert_xaxis()     # Temperature axis is inverted (hotter stars to the left)

    ax.set_xscale('log')  # Temperature axis is logarithmic
    #ax.invert_xaxis()     # Temperature axis is inverted (hotter stars to the left)
    
    ax.set_title('Hertzsprung-Russell Diagram', fontsize=18)
    ax.set_xlabel('Temperature (K)', fontsize=14)
    ax.set_ylabel('Luminosity (Solar units)', fontsize=14)
    
    ax.grid(True, which='both', linestyle='--', alpha=0.5)

    plt.show()

# --- Example Usage with Sample Data ---
# In a real-world scenario, you would load real stellar data from a catalog.
# For this example, we generate some synthetic data points.
def generate_sample_data(num_points=2000):
    """Generates synthetic stellar data for demonstration purposes."""
    np.random.seed(42)
    # Simulate stars primarily on the main sequence
    temps = np.random.lognormal(mean=np.log(6000), sigma=0.5, size=num_points)
    lums = (temps / 5778)**4 * np.random.lognormal(mean=0, sigma=0.4, size=num_points)
    
    # Add a few giants and white dwarfs for diversity
    temps = np.append(temps, np.random.uniform(3000, 6000, 50))
    lums = np.append(lums, np.random.uniform(100, 1000, 50))
    temps = np.append(temps, np.random.uniform(8000, 20000, 50))
    lums = np.append(lums, np.random.uniform(1e-4, 1e-2, 50))

    return {'temperature': temps, 'luminosity': lums}

# Generate and plot the HRD
stellar_data = generate_sample_data()
create_hrd_plot(stellar_data)

```
