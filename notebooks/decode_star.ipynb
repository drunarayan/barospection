### Decode Blind ID and Explore your Star!

```python
def decode_two_integers(encoded_int, bit_shift):
    """
    Decodes a single integer back into two original integers.

    Args:
        encoded_int (int): The encoded integer.
        bit_shift (int): The same bit_shift used during encoding.

    Returns:
        tuple: A tuple containing the two original integers (int1, int2).
    """
    int1 = encoded_int >> bit_shift
    int2 = encoded_int & ((1 << bit_shift) - 1) # Mask to get the lower bits
    return int1, int2

# Example usage:
#num_a = 596
#num_b = 728

shift_amount = 10 # Assuming num_b will fit within 10 bits

def googlesearchurl(tname):
    try:
        from googlesearch import search
    except ImportError:
        print("No module named 'google' found")
    #
    # to search
    query = f'{tname} Wikipedia Astronomy'
    
    #for j in search(query, tld="co.us", num=10, stop=10, pause=2):
    for url in search(query, num_results=10, lang="en", unique=True):
        print(url)
#
# example
#
#tname = f"HD{decoded_a}"
#googlesearchurl(tname)
```

```python
import re
while True:
    rbid = input("\nEnter (copy-paste) BID or 'q' to quit): ")
    if rbid.lower() == 'q':  # .lower() makes it case-insensitive
        break  # Exit the loop if 'q' is entered
    match = re.search(r"\d+", rbid)
    if (match):
        bid = int(match.group(0))
        print(f"The Blind ID is: {bid}")
                  
        hd, x, = decode_two_integers(int(bid), shift_amount)
        print(f"The HD Catalog number for Star is: {hd}")
        print(f"\nExplore your star using links below for Star: HD{hd}\n")
        tname = f"HD{hd}"
        
        googlesearchurl(tname)
    else: print("no BID number entered, try again, or 'q' to quit")

print("decode ended ...")
```
